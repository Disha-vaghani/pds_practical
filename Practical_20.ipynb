```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

names = ["A", "B", "C", "D", "E"]
values = [20, 35, 25, 40, 30]

data = [10, 20, 20, 25, 30, 30, 30, 35, 40, 45]

x_scatter = [1, 2, 3, 4, 5]
y_scatter = [2, 4, 5, 8, 10]
```

```python
plt.figure(figsize=(6, 4))

plt.plot(x, y, marker="o")

plt.title("Line Plot")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

```python
plt.figure(figsize=(6, 4))

plt.bar(names, values, color="green")

plt.title("Bar Chart")
plt.xlabel("Names")
plt.ylabel("Values")

plt.show()
```

```python
plt.figure(figsize=(6, 4))

plt.hist(data, color="orange", edgecolor="black")

plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(6, 4))

plt.pie(values, labels=names, autopct="%1.1f%%")

plt.title("Pie Chart")

plt.show()
```

```python
plt.figure(figsize=(6, 4))

plt.scatter(x_scatter, y_scatter, color="red")

plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```
