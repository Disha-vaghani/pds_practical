```python
def f4(c,e):
    print("the value of c",c,"the value of e",e)
    print("addition of c and e",c+e)
    print("subtraction of c and c",c-e)
    print("division of c and e",c/e)
    print("multiplication of c and e",c*e)
```

```python
f4(c=2,e=2)
```

```python
#function with return type and argument
def f4(c,e):
    return c+e,c-e,c/e,c*e
```

```python
f4(c=2,e=2)
```
