```python
#practical 6

# try and except using exception handling

x = 10
try:
    print(x)
except NameError:
    print("Variable x is not defined")
except Exception:
    print("Something else went wrong")
```

```python
#else using exception handling

try:
    print("Hello")
except:
    print("Something went wrong")
else:
    print("Nothing went wrong")
```

```python
#finally using exception handling

try:
    print(x)
except:
    print("Something went wrong")
finally:
    print("The 'try except' is finished")
```
