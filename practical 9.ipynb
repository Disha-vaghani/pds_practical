```python
#practical 9
import pandas as pd
import numpy as np

# Read Excel file
df = pd.read_excel(r"C:\Users\Students\Downloads\employee.xlsx")

# Convert Excel to CSV
df.to_csv(r"C:\Users\Students\Downloads\employee.csv", index=False)

print(df.head())
print("Excel file converted to CSV successfully!")
```

```python
print(df["DailyRate"].describe())
```

```python
import matplotlib.pyplot as plt

plt.hist(df["DailyRate"], bins=10)

plt.xlabel("DailyRate")
plt.ylabel("Number of Employees")
plt.title("Distribution of Employee DailyRate")

plt.show()
```

```python
plt.scatter(df["DailyRate"], df["MonthlyIncome"])

plt.xlabel("DailyRate")
plt.ylabel("Monthly Income")
plt.title("Age vs Monthly Income")

plt.show()
```

```python
correlation = df["DailyRate"].corr(df["MonthlyIncome"])

print("Correlation =", correlation)
```

```python
variables = ["DailyRate","MonthlyIncome","YearsAtCompany","TotalWorkingYears"]

correlation_matrix = df[variables].corr()

print(correlation_matrix.round(2))
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="Blues",
    fmt=".1f",
    linewidths=1.0,
    linecolor="black",
    square=True,
    cbar_kws={"shrink": 0.6}
)
plt.title("Correlation Matrix of Employee Variables")
plt.show()
```
