```python
#practical 3

n=int(input("Enter a number: "))

# if-else
if n%2==0:
    print("Even")
else:
    print("Odd")
```

```python
#elif
n=int(input("enter a number: "))

if n>0:
    print("Positive")
elif n<0:
    print("Negative")
else:
    print("Zero")
```

```python
#nested if
n=int(input("enter a number: "))

if n>=0:
    if n%2==0:
        print("Positive Even")
    else:
        print("Positive Odd")

```
