```python
from ucimlrepo import fetch_ucirepo 

# fetch dataset 
rice_cammeo_and_osmancik = fetch_ucirepo(id=545) 
  
# data (as pandas dataframes) 
X = rice_cammeo_and_osmancik.data.features 
y = rice_cammeo_and_osmancik.data.targets 
```

```python
import pandas as pd
import numpy as np
```

```python
df = pd.concat([X, y], axis=1)
```

```python
df.head()
```

```python
# Slicing with .iloc and .loc
print('Positional Slicing (.iloc):\n', df.iloc[0:3, 0:4])
print('\nLabel Slicing (.loc):\n', df.loc[0:2, ['Area', 'Perimeter', 'Major_Axis_Length']])
```

```python
# Filtering
filtered = df[(df['Major_Axis_Length'] > 200) & (df['Minor_Axis_Length'] >= 87)]
print(f'\nFiltered Count (Major_Axis_Length > 200 & Minor_Axis_Length >= 87): {len(filtered)}')
```

```python
print('Sorted by Age (Descending):\n', df[['Area', 'Perimeter', 'Eccentricity']].sort_values(by='Area', ascending=True).head())
```

```python
# Concatenating (Vertical and Horizontal)
v_concat = pd.concat([df.iloc[:2, :3], df.iloc[2:4, :3]], axis=0)
print('\nVertical Concatenation:\n', v_concat)

h_concat = pd.concat([df[['Area']].head(), df[['Perimeter']].head()], axis=1)
print('\nHorizontal Concatenation:\n', h_concat)
```

```python
# Aggregating with groupby
print('Average Eccentricity by Perimeter:\n', df.groupby('Perimeter')['Eccentricity'].mean().head(4))
```

```python
# Normalizing Data (Min-Max Scaling)
df['Area_norm'] = (df['Area'] - df['Area'].min()) / (df['Area'].max() - df['Area'].min())
print('\nOriginal vs Min-Max Normalized Area:\n', df[['Area', 'Area_norm']].head(3))
```
