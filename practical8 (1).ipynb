```python
DailyRate = df["DailyRate"]
```

```python
mean = DailyRate.mean()

print("Mean =", mean)
```

```python
median = DailyRate.median()

print("Median =", median)
```

```python
mode = DailyRate.mode()

print("Mode =", mode.tolist())
```

```python
variance = DailyRate.var()

print("Variance =", variance)
```

```python
population_variance = DailyRate.var(ddof=0)

print(population_variance)
```

```python
std = DailyRate.std()

print("Standard Deviation =", std)
```

```python
minimum =DailyRate.min()
maximum = DailyRate.max()

range_value = maximum - minimum

print("Minimum =", minimum)
print("Maximum =", maximum)
print("Range =", range_value)

# The employee ages range from 18 to 60 years, giving a total range of 42 years.
```

```python
Q1 =DailyRate.quantile(0.25)
Q2 = DailyRate.quantile(0.50)
Q3 = DailyRate.quantile(0.75)

print("Q1 =", Q1)
print("Q2 =", Q2)
print("Q3 =", Q3)
```

```python
IQR = Q3 - Q1

print("IQR =", IQR)
```
