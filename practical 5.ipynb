```python
# practical 5
# Write Mode using text
file = open("student.txt", "w")
file.write("Name: Rahul\n")
file.write("Marks: 90")
file.close()
print("Data written successfully.")
```

```python
#read mode using text
file = open("student.txt", "r")
content = file.read()
print(content)
file.close()
```

```python
#append mode using text
file = open("student.txt", "a")
file.write("\nName: Priya")
file.write("\nMarks: 95")
file.close()
print("data appended successfully.")
```

```python
#write mode using csv
import csv
file = open("student.csv","w",newline="")
writer = csv.writer(file)
writer.writerow(["Name","Marks"])
writer.writerow(["Rahul",90])
file.close()
```

```python
#read mode using csv
import csv 
file = open("student.csv","r") 
reader = csv.reader(file) 
for row in reader: 
	print(row) 
file.close()
```

```python
#append mode using csv
import csv
file = open("student.csv","w",newline="")
writer = csv.writer(file)
writer.writerow(["priya","90"])
writer.writerow(["Rahul",90])
file.close()
print("data append successfully")
```

```python
# copy image in binary 
source = open("panda.jpg", "rb")
destination = open("panda_copy.jpg", "wb")
destination.write(source.read())
source.close()
destination.close()
print("Image copied successfully.")
```
